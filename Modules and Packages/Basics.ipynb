## Modules And Packages

```python
import math

math.sqrt(15)
```

```python
from math import sqrt,pi # to import some functions from a package
print(sqrt(16))
print(pi)
```

```python
import numpy as np # np is a alias
np.array([1,2,3,4])
```

```python
#To import every module from a package

from math import *
print(pi)
sqrt(16)
```

```python
from package.maths import addition

addition(2,3)
```

```python
from package import maths

maths.addition(2,3)
```

```python
from package.subpackage import mult

mult.multiply(2,3)
```
