```python
lst=[]
for i in range(10):
    lst.append(i**2)
print(lst)
```

```python
my_lst=[x**2 for x in range(10)]
print(my_lst)
```

```python
my_list=[x**2 for x in range(10) if x%2==0]
print(my_list)
```

```python
lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[[i,j] for i in lst1 for j in lst2]
print(pair)
```

```python
my_lst=[x**2 if x%2==0 else -1 for x in range(10)]
print(my_lst)
```
