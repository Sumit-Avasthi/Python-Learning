### Tuple

```python
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
tpl = tuple()
print(type(tpl))
```

```python
#list can be converted into tuples and vice versa
numbers = tuple([1,2,3,4,5,6])
print(numbers)
```

```python
# in tuples you can also store different data items
num = (1,1.34,True,"sumit")
print(num)
```

```python
# Accessing elements in tuples
print(numbers[0],numbers[1])
print(numbers[-1],numbers[-2])
print(numbers[:])
```

```python
#Tuples operation
concat_tuples = numbers+num
print(concat_tuples)
```

```python
print(num*3)
```

```python
# Immutable nature of Tuples

print(numbers)
numbers[1] = 3
```

```python
# Tuples Method

print(numbers.count(1)) #Frequency of 1
print(numbers.index(3)) #What is the index of 3
```

```python
# Packing and Unpacking tuples

packed_tuple = 1,"hello",3.14
print(packed_tuple)


a,b,c = packed_tuple
print(a)
print(b)
print(c)


# Unpacking with *

numbers = (1,2,3,4,5)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
#Nested Tuples

tpl = ((1,2,3,4),(5,6,7,8),("Hello",10,11,12))
print(tpl)

print(tpl[0][3])
```

```python
for el in tpl:
    print(el)
```
