### Dictionary in python

```python
## Creating Dictionary

my_dict = {}
print(type(my_dict))

my_dict = dict()
print(type(my_dict))
```

```python
student={'name':'sumit','age':20,'grade':16}
print(student)
```

```python
student= {"name":"sumit","age":20,"name":24}
student
```

```python
#Accessing Dictionary elements
student={'name':'sumit','age':20,'grade':16}


print(student['name'])
print(student['age'])
print(student['last_name']) #Error
```

```python
print(student.get('name'))
print(student.get('age'))
print(student.get('last_name'))
print(student.get('last_name','Key Not Available'))
```

```python
## Adding a new Key and Updating Key

student['age'] = 21
print(student)
student['new_Key'] = "Value"
print(student)
```

```python
# Delete a Key Value pair

del student['age']
print(student)
```

```python
## Dictionary Methods

keys=student.keys()
print(keys)
values=student.values()
print(values)

items = student.items()
print(items)
```

```python
##Shalow Copy

student={'name':'sumit','age':20,'grade':16}
student_copy=student # Shalow Copy(By reference)
print(student)
print(student_copy)

student['name']="sumit1"
print(student)
print(student_copy)

```

```python
# Deep Copy
student_copy=student.copy()
print(student)
print(student_copy)

student["name"]="sumit2"
print(student)
print(student_copy)
```

```python
#Iterating Over Dictionary

## Using for loop we can iterate over dictionary , keys,values,items

#Keys
for keys in student.keys():
    print(keys)


print()
print()

#Values
for value in student.values():
    print(value)

print()
print()

for item in student.items():
    print(item)


print()
print()
#items wiht tuple unpacking
for key,value in student.items():
    print(key,value)
```

```python
# Nested Dictionary

student={
    "student1":{
        "name":"Sumit Avasthi",
        "age" : 20,
        "grade" : "A",
    },
    "student2":{
        "name":"Peter Parker",
        "age": 25,
        "grade": "A+",
    }
}

print(student)
print(student["student1"]["name"])
print(student["student2"]["name"])
print(student["student2"]["age"])
```

```python
# Iteration Over Nested Dictionary

for student_id,student_info in student.items():
    print(f"{student_id} :-")
    for key,val in student_info.items():
        print(f"{key} : {val}")
    print()
```

```python
# Dictionary Comprehension

squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## With condition
squares={x:x**2 for x in range(5) if x%2==0}
squares
```
