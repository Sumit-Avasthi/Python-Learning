### Operators

```python
## Arithmetic Operators

a=10
b=2

add_result=a+b
sub_result=a-b
float_division_result=a/b
floor_division_result=a//b
mul_result=a*b
expo_result=a**b
modu_result=a%b

print(add_result)
print(sub_result)
print(float_division_result)
print(floor_division_result)
print(mul_result)
print(expo_result)
print(modu_result)
```

```python
## Comparison Operator

a=10
b=10

print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
print(a>b)
print(a<b)
```

```python
## Logical Operator

x=True
y=False

print(x and y)
print(x or y)
print(not(y))
```
