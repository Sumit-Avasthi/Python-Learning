```python
## Why functions
num=24
if num%2 == 0:
    print("Number is Even")
else:
    print("Number is Odd")
```

```python
def even_or_odd(num):
    """Checks even and add"""
    if num%2==0:
        print("Number is Even")
    else:
        print("Number is Odd")
```

```python
even_or_odd(23)
even_or_odd(60)
```

```python
def add(a,b):
    return a+b

print(add(3,4))
print(add(2,4))
```

```python
##Variable Length Arguments
## Positional and KeyWords Argument

## Positional Arguments
def print_val(*args): #Tuple
    print(args)

print_val(1,2,3,"Hellow",3.45)
```

```python
#Keywords Argument
def print_kw(**kwargs):
    print(kwargs)


print_kw(name="sumit avasthi",age=20,grade="A+")
```

```python
def both_types(*args,**kwargs):
    for val in args:
        print(val)
    
    for key,val in kwargs.items():
        print(f"{key}:{val}")


both_types(1,2,3,4,"Hello",name="sumit",age=23)
```
