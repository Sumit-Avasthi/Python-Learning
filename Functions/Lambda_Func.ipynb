## Lambda Function

```python
def Sum(a,b):
    return a+b
```

```python
print(type(Sum))
print(Sum(2,3))
```

```python
addition=lambda a,b:a+b  ## Lambda Function
print(type(addition))
```

```python
print(addition(2,3))
```

```python
check_even=lambda num:num%2==0
```

```python
print(check_even(23))
print(check_even(24))
```
