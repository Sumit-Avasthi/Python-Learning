## Iterators

```python
my_list = [1,2,3,4,5,6]
```

```python
print(type(my_list))
for i in my_list:
    print(i,end=" ")
```

```python
iterator = iter(my_list)
print(type(iterator))
```

```python
print(iterator) ## Lazy loading technique, we have to iterate for elements
```

```python
next(iterator)
```
