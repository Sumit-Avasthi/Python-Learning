### Function Copy

```python
def welcome():
    return "Welcome to our cafe"
```

```python
print(welcome())
```

```python
wel = welcome
print(wel)
wel()
```

```python
print(wel())
del welcome
print(wel())
```

### Closure

```python
## Function inside the function

def main_welcome(msg):
    wel = "Welcome"

    def sub_welcome():
        print("Welcome to our python course")
        print(msg) # We can access the variables of main method inside the sub methods
        print("Learn all the concepts very carefully")

    return sub_welcome()
```

```python
main_welcome("Sumit baba ki jai ho")
```

```python
def main_welcome(func):
    wel = "Welcome"

    def sub_welcome():
        print("Welcome to our python course")
        func("Print function call")
        print("Learn all the concepts very carefully")

    return sub_welcome()
```

```python
main_welcome(print)
```

```python
def main_welcome(func,lst):
    wel = "Welcome"

    def sub_welcome():
        print("Welcome to our python course")
        print(func(lst))
        print("Learn all the concepts very carefully")
    
    return sub_welcome()
```

```python
main_welcome(len,[1,2,3,4,5])
```

## Decorators

```python
def main_welcome(func):
    wel = "Welcome"

    def sub_welcome():
        print("Welcome to our python course")
        func()
        print("Learn all the concepts very carefully")

    return sub_welcome()
```

```python
def decorator():
    print("Hey I am a decorator")
```

```python
decorator()
main_welcome(decorator)
```

```python
@main_welcome
def decorator():
    print("Hey I am a decorator")
```

```python
def my_decorator(func):
    def wrapper():
        print("something before")
        func()
        print("Something after")
    return wrapper
```

```python
def greet():
    print("Good Morning")
```

```python
@my_decorator
def greet():
    print("Good Morning")
```

```python
greet()
```
