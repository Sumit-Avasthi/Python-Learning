## Generators

```python
def square(n):
    for i in range(n):
        return i**2
```

```python
print(square(3))
```

```python
def square(n):
    for i in range(n):
        yield i**2
```

```python
for i in square(3):
    print(i)
```

```python
iterator = square(3)
```

```python
next(iterator)
```

```python
## One of the great example of generators is reading large file

def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
for line in read_large_file('larget_file.txt'):
    print(line.strip()) # to strip new line
```
