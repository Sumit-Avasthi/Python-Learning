## File Handling

```python
f=open('test.txt','r') # Return a file object
```

```python
print(f.read())
f.close()
```

```python
with open('test.txt','r') as f:
    content=f.read()
    print(content)
```

```python
## To read a file line by line

with open('test.txt','r') as file:
    for line in file:
        print(line)
```

```python
with open('test.txt','r') as file:
    for line in file:
        print(line.strip()) # To remove the new line character from every line end
```

```python
## The content is overwritten with 'w' mode 

with open('test.txt','w') as file:
    file.write('This is the first line\n')
    file.write('This is the new line')
```

```python
## To write in a file without overwritten

with open('test.txt','a') as file:
    file.write('This is the append operation\n')
```

```python
# To write list of lines

lines=["Line one\n","Line two\n","Line three\n"]
with open('test.txt','a') as file:
    file.writelines(lines)
```

## Binary Files

```python
data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
with open('example.bin','rb') as file:
    print(file.read())
```
