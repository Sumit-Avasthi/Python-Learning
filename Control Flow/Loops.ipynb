```python
count=0
while count<=5:
    print(count)
    count += 1
```

### Loop Controll Statements

```python
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
for i in range(10):
    if i%2==0:
        continue
    print(i)
    
```

```python
for i in range(3):
    for j in range(2):
        print(f"i={i} and j={j}")
```

```python
n=int(input())
sum=0
i=1
while i<=n:
    sum += i
    i+=1
print(f"sum={sum}")
```

```python
for i in range(2,100):
    is_prime=True
    for j in range(2,i):
        if i%j==0:
            is_prime=False
            break
    if is_prime:
        print(i)
```
