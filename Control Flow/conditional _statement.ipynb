### Conditional Statements

```python
## if statement

age=20
if age>=18:
    print("You Can Vote")
```

```python
## else statement

age=10

if age>=18:
    print("You can vote")
else:
    print("You can not vote")
```

```python
n=int(input())

if n%2==0:
    if n>0:
        print("Number is positive even")
    elif n==0:
        print("Number is negative even")
    else:
        print("Number is zero")
else:
    if n>0:
        print("Number is positive odd")
    elif n==0:
        print("Number is negative odd")
    else:
        print("Number is zero")
```

```python

year=int(input())
if (year%4==0 and year%100 !=0) or (year%400 == 0):
    print("Its a leap year")
else :
    print("its not a leap year")
```

```python
## Simple calculator

a=int(input())
b=int(input())

c=input()

if c=="+":
    print(a+b)
elif c=="-":
    print(a-b)
elif c=="*":
    print(a*b)
elif c=="/":
    print(a/b)
else:
    print("invalid operation")
```
