```python
#2-way : Best practice

class Dog:
    #constructor - __init__() method
    # self is used to access instance variable inside the class
    def __init__(self,name,age):
        self.name=name # class name = name parameter
        self.age=age #class age = age parameter

dog1=Dog("Buddy",4)
print(dog1.name)
print(dog1.age)


dog2=Dog("Lucy",3)
print(dog2.name)
print(dog2.age)


#if i have to define new attribute for objcts i can do directly in __init__() function
```

```python
# instance method - functionalities of a object 

class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    
    #Instance method
    def Bark(self): # we use self keyword to use instance variables of the class
        print(f"{self.name} says woof woof!!")


dog1 = Dog("Buddy",4)
dog1.Bark()

dog2= Dog("Lucy",3)
dog2.Bark()
```

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance
    
    def deposite(self,money):
        self.balance += money
    
    def widthraw(self,amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Dear {self.owner} amount : {amount} is debited from your account")
        else:
            print(f"{self.owner} insufficient balance")
    
    def owner_details(self):
        print(f"Owner name = {self.owner} \nBalance = {self.balance}")


customer1 = BankAccount("Sumit Avasthi",25000)
customer2 = BankAccount("Suryansh shukla",0)

customer1.owner_details()
customer2.owner_details()

customer1.widthraw(10000)
customer2.widthraw(10000)
```
