### Encapsulation

```python
## public,protected,private access modifier

class Person:
    def __init__(self,name,age):
        self.name = name   #Public variable
        self.age = age     #public variable

#Public variable are accessible outside the class

person=Person("sumit",20)
print(person.name)

dir(person) # Able to see our public variable 
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.__name = name   #Private variable
        self.__age = age     #Private variable
        self.gender=gender   #Public variable
        
```

```python
# Private varibales are not displayed or accessed outside the class even in derived class , we use getter and setter to access it

person = Person("sumit",20,"Male")
print(person.__name)
```

```python
dir(person)

# __name is converted into _Person__name
# __age is converted into _Person_age
```

```python
print(person._Person__age)

#But the user can not see or know this
```

```python
class Person:
    def __init__(self,name,age,gender):
        self._name = name   #Protected variable
        self._age = age     #Protected variable
        self.gender=gender   #Public variable
# We can not access it outside the class but we can using derived class(child class)

class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)

employee = Employee("sumit",20,"Male")
print(employee._name)
```

```python
# Encapsulation with getter and setter

class Person:
    def __init__(self,name,age):
        self.__name = name
        self.__age = age
    
    #Getter for name private variable
    def get_name(self):
        return self.__name
    
    #Setter for name private variable
    def set_name(self,name):
        self.__name = name
    

person = Person("Sumit",21)

person.get_name() ## To access the private variable of class
```
