```python
try:
    num=int(input("Enter a number"))
    ans=12/num
except ValueError :
    print("This is not a number")
except ZeroDivisionError :
    print("Enter a valid denominator")
except Exception as ex:#To handle any other expetion
    print(ex)
```

```python
# Try,except , else block
try:
    num=int(input("Enter a number"))
    ans=12/num
except ValueError :
    print("This is not a number")
except ZeroDivisionError :
    print("Enter a valid denominator")
except Exception as ex:#To handle any other expetion
    print(ex)
else :   #If no exception is happend then only it get executed
    print(ans)
```

```python
# try,except,else and finally block

try:
    num=int(input("Enter a number"))
    ans=12/num
except ValueError :
    print("This is not a number")
except ZeroDivisionError :
    print("Enter a valid denominator")
except Exception as ex:#To handle any other expetion
    print(ex)
else :   #If no exception is happend then only it get executed
    print(ans)
finally: #No matter what this block has to be executed
    print("Finally block")
```
